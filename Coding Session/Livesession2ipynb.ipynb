```python
def relu(x):
  return np.maximum(x,0)
```

## Sigmoid

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
sigmoid(x)
```

```python
y = np.arange(-10,10,0.1)
```

```python
import matplotlib.pyplot as plt
```

```python
z = sigmoid(y)
z
```

```python
plt.plot(y,z,label='sigmoid')
plt.title('Sigmoid')
plt.xlabel("x")
plt.ylabel("σ(x)")
plt.grid(True)
plt.legend()
plt.show()
```

### Properties of `Sigmoid`
* Strictly increasing  
* Lies between `[0,1]`

## Softmax

```python
def softmax(x):
  return np.exp(x)/np.sum(np.exp(x))
```
